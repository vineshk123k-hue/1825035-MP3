```python
#FINITE DIFFERENCE METHOD

import numpy as np

# Boundary conditions
a = 0
b = 1
ya = 0
yb = 0

# Number of intervals
n = 10
h = (b - a) / n

# Grid points
x = np.linspace(a, b, n + 1)

# Coefficient matrix and RHS
A = np.zeros((n-1, n-1))
B = np.zeros(n-1)

for i in range(n-1):
    A[i, i] = -2

    if i > 0:
        A[i, i-1] = 1

    if i < n-2:
        A[i, i+1] = 1

    B[i] = -2 * h**2

# Apply boundary conditions
B[0] -= ya
B[-1] -= yb

# Solve the equations
y = np.linalg.solve(A, B)

# Add boundary values
y = np.concatenate(([ya], y, [yb]))

# Display results
print(" x       y")
for i in range(n+1):
    print(f"{x[i]:.2f}   {y[i]:.4f}")
```

```python
#FOURIER TRANSFORM

import numpy as np
import matplotlib.pyplot as plt

# Input signal
N = 100
t = np.linspace(0, 1, N)
f = 5

# Generate sine wave
signal = np.sin(2 * np.pi * f * t)

# Calculate Fourier Transform
fft_result = np.fft.fft(signal)

# Calculate frequency
frequency = np.fft.fftfreq(N, t[1] - t[0])

# Plot original signal
plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title("Original Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")

# Plot Fourier Transform
plt.subplot(2, 1, 2)
plt.plot(frequency[:N//2],
         np.abs(fft_result[:N//2]))
plt.title("Fourier Transform")
plt.xlabel("Frequency")
plt.ylabel("Magnitude")

plt.tight_layout()
plt.show()
```

```python
#ROTATION MATRIX

import numpy as np
import matplotlib.pyplot as plt

x, y = 2, 3
theta = 45

a = np.radians(theta)

R = np.array([[np.cos(a), -np.sin(a)],
              [np.sin(a), np.cos(a)]])

P = np.array([x, y])
Q = R @ P

print("Original Point:", P)
print("Rotated Point:", Q)

plt.scatter(x, y, label="Original")
plt.scatter(Q[0], Q[1], label="Rotated")
plt.grid()
plt.legend()
plt.show()
```

```python
#TRANSLATION
import numpy as np
import matplotlib.pyplot as plt

x, y = 2, 3
tx, ty = 4, 2

P = np.array([x, y])
T = np.array([tx, ty])

Q = P + T

print("Original Point:", P)
print("Translated Point:", Q)

plt.scatter(x, y, label="Original")
plt.scatter(Q[0], Q[1], label="Translated")
plt.grid()
plt.legend()
plt.show()
```

```python
#SCALING
import numpy as np
import matplotlib.pyplot as plt

x, y = 2, 3
sx, sy = 2, 3

P = np.array([x, y])

S = np.array([[sx, 0],
              [0, sy]])

Q = S @ P

print("Original Point:", P)
print("Scaled Point:", Q)

plt.scatter(x, y, label="Original")
plt.scatter(Q[0], Q[1], label="Scaled")
plt.grid()
plt.legend()
plt.show()
```

```python
#Reflection
import numpy as np
import matplotlib.pyplot as plt

x, y = 2, 3

P = np.array([x, y])

R = np.array([[1, 0],
              [0, -1]])

Q = R @ P

print("Original Point:", P)
print("Reflected Point:", Q)

plt.scatter(x, y, label="Original")
plt.scatter(Q[0], Q[1], label="Reflected")
plt.grid()
plt.legend()
plt.show()
```

```python
#IFT

import numpy as np
import matplotlib.pyplot as plt

# Number of samples
N = 100

# Time array
t = np.linspace(0, 1, N)

# Original signal
f = 5
signal = np.sin(2 * np.pi * f * t)

# Fourier Transform
F = np.fft.fft(signal)

# Inverse Fourier Transform
reconstructed_signal = np.fft.ifft(F)

# Display results
print("Original Signal:")
print(signal)

print("\nReconstructed Signal:")
print(np.real(reconstructed_signal))

# Plot original signal
plt.plot(t, signal, label="Original Signal")

# Plot reconstructed signal
plt.plot(t, np.real(reconstructed_signal),
         '--', label="Reconstructed Signal")

plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Inverse Fourier Transform")
plt.legend()
plt.grid()
plt.show()
```

```python
#F(z) vs z

import numpy as np
import matplotlib.pyplot as plt

# Define z values
z = np.linspace(-10, 10, 100)

# Define the function F(z)
F = z**2 + 2*z + 3

# Plot graph
plt.plot(z, F, label="F(z) = z² + 2z + 3")

plt.xlabel("z")
plt.ylabel("F(z)")
plt.title("Graph of F(z) vs z")

plt.grid()
plt.legend()
plt.show()
```

```python
#f(z) vs z where imaginary part is function of real part

import numpy as np
import matplotlib.pyplot as plt

# Real part
x = np.linspace(-5, 5, 100)

# Imaginary part as a function of real part
y = x**2

# Plot
plt.plot(x, y, label="Imaginary part = x²")

plt.xlabel("Real part (x)")
plt.ylabel("Imaginary part (y)")
plt.title("F(z) vs z")

plt.grid()
plt.legend()
plt.show()
```
